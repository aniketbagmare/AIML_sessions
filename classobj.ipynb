```python
# if i have multiple objects from same class
class Employee:
    def __init__(self,first,last,salary):
        self.first = first
        self.last = last
        self.email = '{}{}.gamil.com'.format(self.first,self.last)
        self.salary = salary
        
    def return_full_name(self):
        return '{} {}'.format(self.first,self.last)  
    
emp1 = Employee('Ram','Patil',55000)
emp2 = Employee('Sham','Patel',72000)
    
print(emp1.email)  
print(emp2.email)  

print(emp1.return_full_name())

# how methodes call internally
print(Employee.return_full_name(emp1))

print(emp1.__dict__)
print(emp2.__dict__)

# add new variable in emp2 
emp2.role = 'AI engineer'

print(emp2.__dict__)
```

```python
class Employee:
    def __init__(self,fname,lname,salary):
        self.fname = fname
        self.lname = lname
        self.email = '{}{}@gmail.com'.format(self.fname,self.lname)
        self.salary = salary
        
    def return_name(self):
        return '{} {}'.format(self.fname,self.lname)
    
    def apply_raise(self):
        self.salary = self.salary * 1.04
        
e = Employee('Om','Birla',30000) 
print(e.salary)
e.apply_raise()  
print(e.salary)
e.apply_raise()
print(e.salary)         
```

```python
class Employee:
    raise_amount = 1.04
    
    def __init__(self,fname ,lname,salary):
        self.fname = fname
        self.lname = lname
        self.email = '{}{}@company.in'.format(self.fname,self.lname)
        self.salary = salary
        
    def return_name(self):
        return '{}{}'.format(self.fname,self.lname)
    
    def apply_raise(self) :
        self.salary = self.salary * self.raise_amount  
        
e = Employee('Prakash','Gaitonde',45000)
print(e.salary)
e.apply_raise()
print('after 4%  Increase salary : ',e.salary)
print(Employee.raise_amount)

# print(e.__dict__)
# print(Employee.__dict__)
print('after change raise ammount from 4% to 5% ')
Employee.raise_amount = 1.05
print(Employee.raise_amount)

print(e.salary)
e.apply_raise()
print('salary increase by 5% :',e.salary)
         
```
