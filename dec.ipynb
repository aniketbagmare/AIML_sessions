```python
def print_decorator(func):
    def wrap():
        print('-'*20)
        func()
        print('-'*20)
    return wrap
    
```

```python
@print_decorator
def hello():
    print('hello world')
    
hello()    
```

```python
def xyz(lol):
    def abc():
        print('line one')
        lol()
        print('line two')
    return abc    
        
```

```python
@xyz
def odd():
    print('parle-g')
    
odd()    
```

```python
def name(something):
    def aname():
        print('oooooo')
        something()
        print('pppppp')
    return aname

@name
def pqr():
    print('mango')
    
pqr()    
    
```

```python
def programm(lines):
    def programm2():
        print('nothing')
        lines()

        def programm3():
            print('nothing ' * 3)
            lines()

        programm3()

    return programm2


@programm
def anno():
    print('nothing ' * 2)
    
anno()    
```

```python
# @property decorator
class Employee:
   def __init__(self,fname,lname):
       self.fname = fname 
       self.lname = lname
   
   @property    
   def email(self):
       return '{}{}@gmail.com'.format(self.fname,self.lname)
   
   
e1 = Employee   ('Ram','Patil')
print(e1.fname)
# print(e1.email()) # it is act as a method
print(e1.email) # now it is act like a variable

print('-'*20)
e1.fname = 'Sham'
# print(e1.email())
print(e1.email)
            
```

```python
class Employee:
   def __init__(self,fname,lname):
       self.fname = fname 
       self.lname = lname
   
   @property    
   def email(self):
       return '{}{}@gmail.com'.format(self.fname,self.lname)
   
   @property
   def fullname(self):
       return '{} {}'.format(self.fname ,self.lname)
   
   @fullname.setter
   def fullname(self,name):
       fname , lname = name.split(' ')
       self.fname = fname
       self.lname = lname 
       
emp_2 = Employee('Ram','Patil')
print(emp_2.fullname)
       
emp_2.fullname = 'Sham Jadhav' 
print(emp_2.fullname  )    
```

```python
class Employee:
   def __init__(self,fname,lname):
       self.fname = fname 
       self.lname = lname
   
   @property    
   def email(self):
       return '{}{}@gmail.com'.format(self.fname,self.lname)
   
   @property
   def fullname(self):
       return '{} {}'.format(self.fname ,self.lname)
   
   @fullname.setter
   def fullname(self,name):
       fname , lname  = name.split(' ')
       self.fname = fname
       self.lname = lname 
       
   @fullname.deleter
   def fullname(self):
       print('full name deleted')
       self.fname = None
       self.lname = None    
       
emp_2 = Employee('Ram','Patil')
print(emp_2.fullname)

del emp_2.fullname
print(emp_2.fname)
print(emp_2.fullname)
       
        
       
```
